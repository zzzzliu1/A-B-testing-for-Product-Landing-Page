## Calculate the Sample Size

```python
# we want to have 2% progress of conversion from previous 12% on old page.
effect_size = sms.proportion_effectsize(0.12,0.14)
# 这里的意思好似说current old page conversion rate= 12%， 希望new page至少达到=14%
# 所以Absolute improvement = 14%-12% = 2 percentage point

# power of a test usually set to 0.8
# alpha= 0.05
required_sample = sms.NormalIndPower().solve_power(
    effect_size, 
    power=0.8, 
    alpha=0.05, 
    ratio=1
)


#round it
required_sample = ceil(required_sample)                     

print("Sample needed from each group:", required_sample)


# 为了判断新Landing page 是否真的能把Conversion Rate 从12% 提升到14%，我们的实验至少需要多少用户？
# Effect Size= 我们希望实验有能力检测出来的效果大小
# Minimum Detectable Effect（MDE）：我们认为商业上值得检测出来的最小效果
# Power =80% 说明50% 的概率能够检测到这个效果；反过来说有大约：20% probability of Type II error也就是实际上new page 有效果，但实验没有检测出来
# Alpha=0.05；它是Type I error也就是：实际上new page 没有效果，但我们错误地认为它有效。
Type I error = False positive "其实没效果，但我说有效果”
Type II error= False Negative “其实有效果，但我没发现”
```

#### Sampling from both group and control using fixed random_state

```python
# sampling with fixed random_state so the data to ensure reproducibility
control_new = df_unique[df_unique['group'] == 'control'].sample(n=required_sample, random_state=7)
treatment_new = df_unique[df_unique['group'] == 'treatment'].sample(n=required_sample, random_state=7)

ab_test = pd.concat([control_new, treatment_new], axis=0)
ab_test.reset_index(drop=True, inplace=True)

ab_test
```

## Result

```python
conversion_rates = ab_test.groupby('group')['converted']

# Std. deviation of the proportion
std_p = lambda x: np.std(x)  

# Std. error of the proportion 
se_p = lambda x: stats.sem(x)            

conversion_rates = conversion_rates.agg([np.mean, std_p, se_p])
conversion_rates.columns = ['conversion_rate', 'std_deviation', 'std_error']

conversion_rates


# 从结果来看新页面在这个sample中并没有表现得很好，反而低约1.33pp
# standard deviation: 单个用户的converted outcome 在组内有多大差异，因为用户只有 0/1 所以这是Bernoulli outcome 的dispersion-》 individual observaiton有多分散
# 然后se_p这是standard error：它回答的是我们估计出来的conversion rate有多不确定？
# sampe size越大，我们估计Conversion Rate通常越稳定
```

### Hypothesis Testing

```python
from statsmodels.stats.proportion import proportions_ztest, proportion_confint

# 这里导入了两个函数：
proportions_ztest -> Two-proportion Z-Test
proportio_confit -> Confidence Interval 
为什么使用Two-proportion Z-test？


#select the conversion results for the control group
control_results = ab_test[ab_test['group'] == 'control']['converted']

#select the conversion results for the treatment group
treatment_results = ab_test[ab_test['group'] == 'treatment']['converted']

#get the sample sizes for both groups
n_con = control_results.count()
n_treat = treatment_results.count()

#get the number of conversions for both groups
successes = [control_results.sum(), treatment_results.sum()]

#put the sample sizes and number of conversions into the format required by statsmodels proportions_ztest function
nobs = [n_con, n_treat]

#calculate the z-statistic and p-value
z_stat, pval = proportions_ztest(successes, nobs=nobs)

#calculate the 95% confidence intervals for the conversion rates
(lower_con, lower_treat), (upper_con, upper_treat) = proportion_confint(successes, nobs=nobs, alpha=0.05)

print(f'z statistic: {z_stat:.3f}')
print(f'p-value: {pval:.3f}')
print(f'ci 95% for control group: [{lower_con:.3f}, {upper_con:.3f}]')
print(f'ci 95% for treatment group: [{lower_treat:.3f}, {upper_treat:.3f}]')

# we did not find sufficient statistical evidence of a difference between the two pages at the 5% significance level.
```

## Conclusions

1. The conversion rate for the control group is 12.43% and for the greatment group is 11.10%
2. The z-statistic is 1.94, which is less than the critical value of 1.96 for a significance level of 0.05. Therefore, we fail to reject the null hypothesis, which means that there is no statistically significant difference in conversion rates between the old and new landing pages. 
3. The p-value is 0.052,which is greater than the significance level of 0.05. This confirms that there is not enough evidence to reject the null hypothesis 
4. The 95% confidence intervals for control groups shows that we can be 95% confident that the true conversion rate of the control group is somewhere between 11.5% and 13.4%
5. The 95% confidence intervals for treatment groups does no toverlap with the desired 14% conversion rate, which further supported the conclusion that th enew landing page did not result in a significant improvement in conversion rates.
6.  Overall, the A/B testing did not show a significnat improvement in conversion rates from 12% to 14%, and it is recommended to keep the old landing page.
